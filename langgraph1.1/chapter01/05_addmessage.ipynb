```python
from langchain.messages import  SystemMessage,HumanMessage,AIMessage
from langgraph.graph.message import add_messages
from rich import print

left = [
    SystemMessage(content="あなたはプロの翻訳者です",id = '1'),
    HumanMessage(content="こんにちは",id = '2'),
    AIMessage(content="こんにちは、私はプロの翻訳者です",id = '3'),
]

right = [
    HumanMessage(content="私は田中です、あなたは鈴木です",id = '2'),
    AIMessage(content="わかりました、覚えました",id = '3'),
    HumanMessage(content="あなたは誰ですか?",id = '4'),
    AIMessage(content="私は鈴木です",id = '5'),
    HumanMessage(content="あなたは鈴木ですか?",id = '1'),
]
merged = add_messages(left,right)
print(merged)
```

```python
from langgraph.graph import StateGraph, START, END
from typing import TypedDict

class OverAllState(TypedDict):
    logs: list[str]
    id: str

def node_a(state: OverAllState):
    return {
        "logs": ["node_a"],
        "id": "node_a"
    }

def node_b(state: OverAllState):
    return {
        "logs": ["node_b"],
        "id": "node_b"
    }

builder = StateGraph(state_schema=OverAllState)
builder.add_node("node_a", node_a)
builder.add_node("node_b", node_b)
builder.add_edge(START, "node_a")
builder.add_edge("node_a", "node_b")
builder.add_edge("node_b", END)

graph = builder.compile()
result = graph.invoke({"logs": ["START"], "id": "start"})
print('=' * 30, '-> result <-', '=' * 30)
print(result)
```
