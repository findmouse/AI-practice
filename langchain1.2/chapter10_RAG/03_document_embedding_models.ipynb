# ドキュメント埋め込みモデル

## 1、埋め込みモデルの選定と初期化

### ケース1：CloseAI プラットフォームを使用

```python
import os
from dotenv import load_dotenv
from langchain.embeddings import init_embeddings


load_dotenv(override=True)


embedding_model1 = init_embeddings(
    model="openai:text-embedding-3-large",
    api_key=os.getenv("OPENROUTER_API_KEY"),
    base_url=os.getenv("OPENROUTER_API_BASE"),
)
```

### ケース2：OpenAI プラットフォームを使用

```python
import os
from dotenv import load_dotenv
from langchain_openai import OpenAIEmbeddings

load_dotenv(override=True)


embedding_model2 = OpenAIEmbeddings(
    model="text-embedding-3-large",
    api_key=os.getenv("OPENAI_API_KEY"),
)
```

## 2、文のベクトル化

```python

text = "こんにちは、はじめまして"


embed_docs = embedding_model1.embed_query(text)

print(len(embed_docs))

print(embed_docs[:5])
```

```python

text = "こんにちは、はじめまして"


embed_docs = embedding_model2.embed_query(text)

print(len(embed_docs))

print(embed_docs[:5])
```

## 3、ドキュメントのベクトル化
文のリストと理解できる

例1：

```python
# 埋め込み対象のテキストリスト
texts = [
    "Hi there!",
    "Oh, hello!",
    "What's your name?",
    "My friends call me World",
    "Hello World!"
]

# 埋め込みベクトルを生成
embeded_docs = embedding_model1.embed_documents(texts)

for i in range(len(texts)):
    print(f"{texts[i]}:{embeded_docs[i][:3]}",end="\n\n")
```

例2：

```python
from langchain_community.document_loaders import CSVLoader


# ケース1：
loader = CSVLoader("../asset/load/02-load.csv", encoding="utf-8")
docs = loader.load_and_split()

#print(len(docs))

# 各 chunk の embedding が格納される。

texts = [doc.page_content for doc in docs]

embeded_docs = embedding_model2.embed_documents(texts)

print(len(embeded_docs))

for i in range(len(texts)):
    print(f"{texts[i]}:\n{embeded_docs[i][:3]}",end="\n\n")
```
