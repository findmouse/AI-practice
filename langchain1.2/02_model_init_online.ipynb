## モデル初期化の使用方法
# 1、temperatureの使用


```python
from langchain_community.chat_models import ChatZhipuAI, ChatTongyi
from langchain_deepseek import ChatDeepSeek
import os
from rich import print

from dotenv import load_dotenv
from langchain_openrouter import ChatOpenRouter

load_dotenv(override=True)
from langchain.chat_models import init_chat_model
DEEPSEEK_API_KEY = os.getenv("DEEPSEEK_API_KEY")
DEEPSEEK_BASE_URL = os.getenv("DEEPSEEK_BASE_URL")


llm_deepseek = init_chat_model(
    "deepseek-v4-flash",
    temperature=0,
    api_key = DEEPSEEK_API_KEY
)
for _ in range(3):
    response = llm_deepseek.invoke("春に関する七言絶句を一首書いてください。")
    print(response.content)
```

```python
llm_deepseek = init_chat_model(
    "deepseek-v4-flash",
    temperature=1.9,
    api_key = DEEPSEEK_API_KEY
)
for _ in range(3):
    response = llm_deepseek.invoke("春に関する七言絶句を一首書いてください。")
    print(response.content)
```

```python
from langchain.chat_models import init_chat_model
from dotenv import load_dotenv
import os

DEEPSEEK_BASE_URL = os.getenv("DEEPSEEK_BASE_URL")
DEEPSEEK_MODEL = os.getenv("DEEPSEEK_MODEL")
# .envファイルから環境変数を読み込む
load_dotenv(override=True)
model = init_chat_model(
    model=DEEPSEEK_MODEL,
    model_provider="openai",
    temperature=0,
    api_key=os.getenv("DEEPSEEK_API_KEY"),
    base_url=os.getenv("DEEPSEEK_BASE_URL"),
)
# モデルに1件のデータを送信する
response = model.invoke("張三、男性、30歳、プログラミング開発経験8年、現在大手インターネット企業でテクニカルスペシャリストを務めている。上記の文章からデータを抽出し、JSON形式で返してください")
print(response.content)
```

## クリエイティブ案


```python
from langchain.chat_models import init_chat_model
from dotenv import load_dotenv
import os
# .envファイルから環境変数を読み込む
load_dotenv(override=True)
model = init_chat_model(
    model=DEEPSEEK_MODEL,
    model_provider="openai",
    temperature=1.5,
    api_key=os.getenv("DEEPSEEK_API_KEY"),
    base_url=os.getenv("DEEPSEEK_BASE_URL"),
)
# モデルに1件のデータを送信する
response = model.invoke("究極の静音メカニカルキーボードのために、詩的で力強いキャッチコピーを3つ書いてください。")
print(response.content)
```

```python
from langchain.chat_models import init_chat_model
from dotenv import load_dotenv
import os
# .envファイルから環境変数を読み込む
load_dotenv(override=True)
model = init_chat_model(
    model=DEEPSEEK_MODEL,
    model_provider="openai",
    # temperature=1.5,
    api_key=os.getenv("DEEPSEEK_API_KEY"),
    base_url=os.getenv("DEEPSEEK_BASE_URL"),
    max_tokens=10,
)
# モデルに1件のデータを送信する
response = model.invoke("あなた自身について紹介してください。")
print(response.content)
```
