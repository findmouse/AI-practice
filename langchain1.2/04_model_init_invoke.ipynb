## モデル呼び出しのinvokeの使用について

```python
from langchain_community.chat_models import ChatZhipuAI, ChatTongyi

from langchain_deepseek import ChatDeepSeek
import os

from dotenv import load_dotenv
from langchain_openrouter import ChatOpenRouter

load_dotenv(override=True)
```

```python
from langchain.chat_models import init_chat_model
DEEPSEEK_API_KEY = os.getenv("DEEPSEEK_API_KEY")


llm_deepseek = init_chat_model(
    "deepseek-v4-flash",
    api_key = DEEPSEEK_API_KEY
)


```

```python
response = llm_deepseek.invoke('次の漢字を翻訳してください：你好世界。')
print(response)
```

### 辞書リスト

```python
messages = [
	{"role": "system", "content": "あなたはプロの数学教師です"},
	{"role": "user", "content": "フィボナッチ数列とは何か説明してください"}
]
response = llm_deepseek.invoke(messages)
print(response.content)
```

## マルチターン対話の問題

```python
messages = [
	{"role": "system", "content": "あなたはプロの数学教師です"},
	{"role": "user", "content": "1+2=？"},
	{"role": "assistant", "content": "3"}, # オプション、会話履歴に使用
	{"role": "user", "content": "さっき何を質問しましたか？"}
]
response = llm_deepseek.invoke(messages)
print(response.content)
```

# 記憶関連

```python
messages1 = [
{"role": "system", "content": "あなたはとてもフレンドリーなAIアシスタントです"},
{"role": "user", "content": "こんにちは、私の名前は小明です"},
]
# 1回目の対話
response1 = llm_deepseek.invoke(messages1)
# レスポンスを出力
print(f"AIの返信1：{response1.content}")
messages2 = [
{"role": "user", "content": "私の名前は何ですか？"}
]
# 2回目の対話
response2 = llm_deepseek.invoke(messages2)
# レスポンスを出力
print(f"AIの返信2：{response2.content}")
```

```python
conversation = [
{"role": "system", "content": "あなたはとてもフレンドリーなAIアシスタントです"},
{"role": "user", "content": "こんにちは、私の名前は小明です"},
]
# 1回目の対話
response1 = llm_deepseek.invoke(messages1)

conversation.append({'role':"assistant", 'content':response1.content})
conversation.append({'role':"user", 'content':'私の名前は何ですか？'})

response = llm_deepseek.invoke(conversation)
# レスポンスを出力
print(f"AIの返信2：{response.content}")
```

## メッセージリスト

```python
from langchain_core.messages import HumanMessage, SystemMessage,  AIMessage

conversation = [
    SystemMessage(content="あなたはとてもフレンドリーなAIアシスタントです"),
    HumanMessage(content="こんにちは、私の名前は小明です"),
]
# 1回目の対話
response1 = llm_deepseek.invoke(messages1)

conversation.append(AIMessage(content=response1.content))
conversation.append(HumanMessage(content='私の名前は何ですか？'))

response = llm_deepseek.invoke(conversation)
# レスポンスを出力
print(f"AIの返信2：{response.content}")
```
