# ChatPromptTemplateの使用

## 1、ChatPromptTemplateのインスタンス化

### 1.1 from_messages() --推奨

例：

```python
from langchain_core.messages import HumanMessage
from langchain_core.prompts import ChatPromptTemplate, HumanMessagePromptTemplate

chat_prompt_template = ChatPromptTemplate.from_messages([
    ("system","あなたは親切なAIアシスタントで、名前は{name}です"),
    ("human","こんにちは、最近どうですか？"),
    ("ai","元気です、ありがとう"),
    ("human","{user_input}")
])

# 呼び出し
result = chat_prompt_template.invoke({"name":"シャオジー","user_input":"2 + 2 = ？"})
print(result)
```

## 2、ChatPromptTemplateの呼び出し

### 2.1 invoke()

入力パラメータ：辞書のリスト；戻り値の型：ChatPromptValue

例：

```python
from langchain_core.prompts import ChatPromptTemplate

chat_prompt_template = ChatPromptTemplate.from_messages([
    ("system","あなたは親切なAIアシスタントで、名前は{name}です"),
    ("human","こんにちは、最近どうですか？"),
    ("ai","元気です、ありがとう"),
    ("human","{user_input}")
])

# 呼び出し
result = chat_prompt_template.invoke({"name":"シャオジー","user_input":"2 + 2 = ？"})
print(result)
print(type(result))
```

### 2.2 format()

パラメータの型：変数値；戻り値の型：文字列

例：

```python
from langchain_core.prompts import ChatPromptTemplate

chat_prompt_template = ChatPromptTemplate.from_messages([
    ("system","あなたは親切なAIアシスタントで、名前は{name}です"),
    ("human","こんにちは、最近どうですか？"),
    ("ai","元気です、ありがとう"),
    ("human","{user_input}")
])

# 呼び出し
result = chat_prompt_template.format(name="シャオジー",user_input="2 + 2 = ?")
print(result)
print(type(result))
```

### 2.3 format_messages()

入力パラメータ：変数値；戻り値の型：メッセージのリスト

例：

```python
from langchain_core.prompts import ChatPromptTemplate

chat_prompt_template = ChatPromptTemplate.from_messages([
    ("system","あなたは親切なAIアシスタントで、名前は{name}です"),
    ("human","こんにちは、最近どうですか？"),
    ("ai","元気です、ありがとう"),
    ("human","{user_input}")
])

# 呼び出し
result = chat_prompt_template.format_messages(name="シャオジー",user_input="2 + 2 = ?")
print(result)
print(type(result))
```

## 3、大規模言語モデルと組み合わせた呼び出し

例

```python
from dotenv import load_dotenv
from langchain_core.prompts import ChatPromptTemplate
import os
from langchain.chat_models import init_chat_model

######1、大規模言語モデルを提供#########
load_dotenv(override=True)

model = init_chat_model(
    model="deepseek:deepseek-v4-flash",
    extra_body={"thinking": {"type": "enabled"}},
)


######2、プロンプトテンプレートを提供#########
chat_prompt_template = ChatPromptTemplate.from_messages([
    ("system","あなたは親切なAIアシスタントで、名前は{name}です"),
    ("human","こんにちは、最近どうですか？"),
    ("ai","元気です、ありがとう"),
    ("human","{user_input}")
])

# 呼び出し
prompt_value = chat_prompt_template.invoke({"name":"シャオジー","user_input":"2 + 2 = ？"})

######3、モデルの呼び出し#########
response = model.invoke(prompt_value)
print(response)
```

### 4.5 BaseMessagePromptTemplateパラメータのリスト

```python
from langchain_core.prompts import SystemMessagePromptTemplate,HumanMessagePromptTemplate

system_mess_template = SystemMessagePromptTemplate.from_template("あなたは親切なAIアシスタントです")

human_mess_template = HumanMessagePromptTemplate.from_template("こんにちは、私は{name}です")

chat_prompt_template = ChatPromptTemplate.from_messages([
    system_mess_template,
    human_mess_template
])

chat_prompt_template.invoke({"name":"太郎"})
```

例：総合的な使用

```python
from langchain_core.prompts import (
    ChatPromptTemplate,
    SystemMessagePromptTemplate,
    HumanMessagePromptTemplate,
)
from langchain_core.messages import SystemMessage, HumanMessage

# 例1：BaseMessage（インスタンス化済みのメッセージ）を使用
system_msg = SystemMessage(content="あなたはAIエンジニアです。")
human_msg = HumanMessage(content="こんにちは！")

# 例2：BaseMessagePromptTemplateを使用
system_prompt = SystemMessagePromptTemplate.from_template("あなたは{role}です。")
human_prompt = HumanMessagePromptTemplate.from_template("{user_input}")

# 例3：BaseChatPromptTemplate（ネストされた ChatPromptTemplate）を使用
nested_prompt = ChatPromptTemplate.from_messages([("system", "ネストされたプロンプト")])

prompt = ChatPromptTemplate.from_messages([
    system_msg,     # MessageLike (BaseMessage)
    human_msg,      # MessageLike (BaseMessage)
    system_prompt,  # MessageLike (BaseMessagePromptTemplate)
    human_prompt,   # MessageLike (BaseMessagePromptTemplate)
    nested_prompt,  # MessageLike (BaseChatPromptTemplate)
])

prompt.invoke({"role":"人工知能の専門家","user_input":"大規模言語モデルの活用シーンを紹介してください"})
```
