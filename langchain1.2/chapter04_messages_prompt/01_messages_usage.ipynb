## 5、会話履歴の最適化
例：

```python
def keep_recent_messages(messages,max_pairs = 3):
    """
    直近のN往復分の会話を保持する
    max_pairs : 保持する会話の往復数（1往復 = user + assistant）
    """

    # system メッセージと会話メッセージを分離する
    system_messages = [m for m in messages if m.get("role") == "system"]
    conversation_messages = [m for m in messages if m.get("role") != "system"]

    # 直近のメッセージペアのみを保持する
    recent_messages = conversation_messages[-(max_pairs * 2):]

    # system メッセージと直近のメッセージペアを返す
    return system_messages + recent_messages
```

```python
# 初期化
long_conversation = [
    {"role": "system", "content": "あなたは Python の講師です"}
]

# 第1ラウンド
long_conversation.append({"role": "user", "content": "リストとは何ですか？一言で説明してください"})
r1 = model.invoke(long_conversation)
long_conversation.append({"role": "assistant", "content": r1.content})

# 第2ラウンド
long_conversation.append({"role": "user", "content": "リストとタプルの違いは何ですか？一言で説明してください"})
r2 = model.invoke(long_conversation)
long_conversation.append({"role": "assistant", "content": r2.content})

# 第3ラウンド
long_conversation.append({"role": "user", "content": "辞書とは何ですか？一言で説明してください"})
r3 = model.invoke(long_conversation)
long_conversation.append({"role": "assistant", "content": r3.content})

print(f"元のメッセージ数: {len(long_conversation)}")  # 7

# 最適化：直近2ラウンドのみを保持
optimized = keep_recent_messages(long_conversation, max_pairs=2)

print(f"最適化後のメッセージ数: {len(optimized)}")  # 5
print(f"保持される内容: system + 直近2ラウンドの会話")

# 新しいユーザーの質問を追加
optimized.append({"role": "user", "content": "私の最初の質問は何でしたか？"})
# 最適化後の履歴を使用
response = model.invoke(optimized)
print(f"\nAI の返答: {response.content}")
```

## 6、マルチターン会話チャットボット

```python
from langchain.chat_models import init_chat_model
import os
from dotenv import load_dotenv
load_dotenv(override=True)

# 1. 基本設定
MODEL_NAME = "gpt-5.4-mini"
MAX_PAIRS_HISTORY = 10
EXIT_WORD = "quit"


OPENROUTER_API_KEY = os.getenv("OPENROUTER_API_KEY")
OPENROUTER_API_BASE = os.getenv("OPENROUTER_API_BASE")

# 2. モデルを初期化
model = ChatOpenRouter(
    model="openai/gpt-5.4-mini",
    api_key=OPENROUTER_API_KEY,
    base_url=OPENROUTER_API_BASE,
)
# 3. メッセージリストを維持する
messages = [
    {
        "role": "system",
        "content":"あなたは小谷姉さん（シャオグージエジエ）、尚硅谷教育のデジタル社員であり、忍耐強く親切なAIアシスタントとして、学習に関する質問に答えることができます"
    }
]

print(f"具体的な質問を入力してください。{EXIT_WORD}と入力すると会話が終了します")

i = 1  # 会話のラウンド数を示す
while True:
    print("\n","="*10,f"第{i}ラウンドの会話開始","="*10,"\n")

    user_input = input("入力してください：")

    # 現在の会話を終了するかどうかを判定
    if user_input == EXIT_WORD:
        print("会話は終了しました。またのご利用をお待ちしております")
        break

    # 4. ユーザーの入力をメッセージリストに追加
    messages.append({
        "role": "user",
        "content": user_input
    })

    print("小谷姉さん：",end="",flush=True)

    # 5. AI の返答内容を連結する
    reply_content = ""

    # 6. 履歴メモリを最適化
    memory_messages = keep_recent_messages(messages, max_pairs=MAX_PAIRS_HISTORY)

    # 7. モデルの応答をストリーミング出力
    for chunk in model.stream(memory_messages):
        if chunk.content:
            print(chunk.content,end="",flush=True)
            reply_content += chunk.content

    print("\n","="*10,f"第{i}ラウンドの会話終了","="*10,"\n")

    i += 1

    # 8. モデルの応答をメッセージリストに追加
    messages.append({"role":"assistant", "content": reply_content})
```
