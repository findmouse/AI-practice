# model_kwargs
langchainが直接列挙していないフィールドでも、モデル自体がサポートしていれば、それらのパラメータをmodel_kwargsパラメータの中で宣言できます。

```python
from langchain.chat_models import init_chat_model
from dotenv import load_dotenv
from rich import print as rprint

# .envファイルから環境変数を読み込む
load_dotenv(override=True)
model = init_chat_model(
    model="deepseek:deepseek-v4-flash",
    model_kwargs={"tools":
        [
            {
                "type": "function",
                "function": {
                    "name": "get_weather",
                    "description": "Get weather of a location, the user should supply a location first.",
                    "parameters": {
                        "type": "object",
                        "properties": {
                            "location": {
                                "type": "string",
                                "description": "The city and state, e.g. San Francisco, CA",
                            }
                        },
                        "required": ["location"]
                    },
                }
            },
        ]
    }
)
# モデルに1件のデータを送信
response = model.invoke("こんにちは、今日の東京の天気はどうですか")
# レスポンスを表示
rprint(response)
```

# extra_body はモデルベンダーがOpenAI APIプロトコルを拡張したフィールドを格納する

```python
from langchain.chat_models import init_chat_model
from dotenv import load_dotenv
from rich import print as rprint

# .envファイルから環境変数を読み込む
load_dotenv(override=True)
model = init_chat_model(
    model="deepseek:deepseek-v4-flash",
    extra_body={"thinking": {"type": "enabled"}},
)
# モデルに1件のデータを送信
response = model.invoke("こんにちは、一言で答えてください")
# レスポンスを表示
rprint(response)
```

# config設定
config設定

```python
from langchain.chat_models import init_chat_model
from dotenv import load_dotenv
import os
from rich import print as rprint

# .envファイルから環境変数を読み込む
load_dotenv(override=True)
DEEPSEEK_API_KEY = os.getenv("DEEPSEEK_API_KEY")
DEEPSEEK_BASE_URL = os.getenv("DEEPSEEK_BASE_URL")
# 1. モデルを初期化
model = init_chat_model(
    model="deepseek-v4-flash",
    model_provider="deepseek",
    api_key=DEEPSEEK_API_KEY,
    base_url=DEEPSEEK_BASE_URL,
    temperature=0.2,
    max_tokens=500,
    # 調整可能なパラメータを指定
    configurable_fields=("model", "model_provider", "temperature", "max_tokens"),
)
# 2. config 辞書を準備
config = {
    "run_name": "joke_generation",  # LangSmithではこの実行がこの名前で表示される
    "joke_generation"
    "tags": ["tag1", "tag2"],  # 分類・検索しやすいようにタグを付ける
    "metadata": {"user_id": "123"},  # ユーザーIDを記録
    "configurable": {
        "model": "deepseek-v4-pro",  # モデルパラメータを設定
        "model_provider": "openai",  # モデルプロバイダーパラメータを設定
        "temperature": 0.7,  # 温度パラメータを設定
        "max_tokens": 1000  # 最大トークン数を設定
    }
}
# 3. モデルを呼び出し、configを渡す
response = model.invoke(
    "1 + 2 = ？",
    config=config
)
rprint(response)
```
