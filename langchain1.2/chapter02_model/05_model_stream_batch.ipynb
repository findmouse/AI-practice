```python
from langchain_community.chat_models import ChatZhipuAI, ChatTongyi

from langchain_deepseek import ChatDeepSeek
import os
from rich import print
from dotenv import load_dotenv
from langchain_openrouter import ChatOpenRouter

load_dotenv(override=True)
```

```python
from langchain.chat_models import init_chat_model

DEEPSEEK_API_KEY = os.getenv("DEEPSEEK_API_KEY")
DEEPSEEK_BASE_URL = os.getenv("DEEPSEEK_BASE_URL")
DEEPSEEK_MODEL = os.getenv("DEEPSEEK_MODEL")

model = init_chat_model(
    model = "deepseek-v4-flash",
    api_key=DEEPSEEK_API_KEY,
    base_url = DEEPSEEK_BASE_URL
)
```

```python
for chunk in model.stream("人工知能とは何か説明してください。"):
    print(chunk.text, end="", flush=True)
```

# バッチ呼び出し

```python
messages = [
    "こんにちは、あなたは誰ですか？",
    "2 + 3 * 5 = ?",
    "中国の首都はどこですか？"
]
responses = model.batch(messages)
for response in responses:
    print(response)
```

# 各リクエストが完了次第、他がすべて完了するのを待たずに結果を返す

```python
messages = [
    "こんにちは、あなたは誰ですか？",
    "2 + 3 * 5 = ?",
    "中国の首都はどこですか？"
]
responses = model.batch_as_completed(messages)
for response in responses:
    print(response)
```

## パフォーマンス比較

```python
inputs = [
    "英訳：春が来た",
    "英訳：夏はとても暑い",
    "英訳：秋は落ち葉の季節",
    "英訳：冬は雪が降る"
]
# ✅ バッチ呼び出し（効率的）
import time
start = time.time()
responses = model.batch(inputs)
batch_time = time.time() - start
print("バッチ呼び出しの結果：")
for i, response in enumerate(responses):
    print(f"{i+1}. {response.content}")
print(f"所要時間: {batch_time:.2f}秒\n")
```

## invoke()をループで呼び出す

```python
# ❌ ループ呼び出し（非効率、比較用のみ）
inputs = [
    "英訳：春が来た",
    "英訳：夏はとても暑い",
    "英訳：秋は落ち葉の季節",
    "英訳：冬は雪が降る"
]
start = time.time()
loop_responses = []
for inp in inputs:
    response = model.invoke(inp)
    loop_responses.append(response)
loop_time = time.time() - start
for i, response in enumerate(responses):
    print(f"{i+1}. {response.content}")
print(f"ループ呼び出しの所要時間: {loop_time:.2f}秒")
print(f"バッチ呼び出しによる短縮率: {((loop_time - batch_time) / loop_time * 100):.1f}%")
```
