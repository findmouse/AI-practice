```python
from langchain.chat_models import init_chat_model
import os
from pydantic import BaseModel, Field
from typing import Union
from langchain.agents import create_agent
from langchain.agents.structured_output import ToolStrategy
from langchain.messages import HumanMessage
from rich import print
from dotenv import load_dotenv
from langchain.agents import create_agent
from langchain.tools import tool
from typing import Dict, Any
from rich import print

load_dotenv(override=True)

OPENROUTER_API_KEY = os.getenv("OPENROUTER_API_KEY")
OPENROUTER_API_BASE = os.getenv("OPENROUTER_API_BASE")

model = init_chat_model(
    model="openai/gpt-5.4-mini",
    model_provider="openai",
    api_key=OPENROUTER_API_KEY,
    base_url=OPENROUTER_API_BASE,
)
```

stream_mode="values"

stream_mode を values モードに設定すると、各ステップの実行後に完全な状態情報が出力されます。各ステップで
完全な状態を取得する必要がある場合や、状態の永続化シナリオに適しています。

```python
@tool
def query_customer_data(customer_id: str) -> Dict[str, Any]:
    """
    顧客の基本情報を照会する
    
    Args:
        customer_id: 顧客ID、顧客を一意に識別するために使用
    
    Returns:
        氏名、等級、加入日など、顧客の基本情報を含む辞書
    """
    # データベース検索をシミュレート
    return {"name": "田中", "level": "VIP", "join_date": "2023-01-15"}


@tool
def check_order_history(customer_id: str) -> Dict[str, Any]:
    """
    顧客の注文履歴を照会する
    
    Args:
        customer_id: 顧客ID、顧客を一意に識別するために使用
    
    Returns:
        総注文数、総支出額など、顧客の注文履歴を含む辞書
    """
    return {"total_orders": 15, "total_spent": 25800.00}


@tool
def get_current_promotions() -> Dict[str, Any]:
    """
    現在利用可能なプロモーション情報を取得する
    
    Returns:
        キャンペーン名、有効期限など、現在利用可能なプロモーション情報を含む辞書
    """
    return {
        "promotions": ["既存顧客優待", "会員限定割引"],
        "valid_until": "2027-01-31"
    }


# カスタマーサービスAgentを作成
customer_service_agent = create_agent(
    model=model,
    tools=[query_customer_data, check_order_history, get_current_promotions])

for chunk in customer_service_agent.stream(
        {"messages": [{"role": "user", "content": "顧客ID CUST123456 の個人情報、注文履歴、利用可能な優待を照会してください"}]},
        stream_mode="values"):
    print(chunk)
    # print(chunk['messages'][0].content, end="", flush=True)
    print("-" * 50)
```

チェックポイント（checkpoint）が作成されるたびに出力がトリガーされ、出力にはチェックポイント内の状態が含まれます。状態の
永続化、ワークフローの復元、または分散実行のトラッキングが必要な高度なシナリオで使用されます。checkpoints

```python
from langgraph.checkpoint.memory import InMemorySaver

# その他のツールコードは上記と同じ、変更なし
# ... ...
# 1. インメモリチェックポイントストレージを作成
checkpointer = InMemorySaver()
# 2. Agentを作成
customer_service_agent = create_agent(
    model=model,
    tools=[query_customer_data, check_order_history,
           get_current_promotions],
    checkpointer=checkpointer  # チェックポイントを有効化
)
# 3. 一意のセッションIDを作成
config = {"configurable": {"thread_id": "session01"}}
# 4. Agentを呼び出す
checkpoint_count = 0
# checkpoints モードでストリーミング監視を行う
for chunk in customer_service_agent.stream(
        {"messages": [{"role": "user", "content": "顧客ID CUST123456 の完全な情報と利用可能な優待を照会してください"}]},
        config=config,
        stream_mode="checkpoints"
):
    checkpoint_count += 1
    print(f"チェックポイント #{checkpoint_count}")
    print(chunk)
    print("-" * 50)
```

customの出力モード    開発者は get_stream_writer を使って、ツールやノードの内部で 送信するデータをカスタマイズ できます。ビジネスロジックに関連する進捗情報（例：“10/100件のレコードを処理済み”）、カスタムログや指標の出力に使用されます。

```python
from langchain.agents import create_agent
from langgraph.config import get_stream_writer
from langchain.tools import tool
import time


@tool
def generate_sales_report() -> str:
    """売上レポートを生成する"""
    writer = get_stream_writer()
    writer({"type": "売上レポート生成", "message": "売上レポートの生成を開始します"})
    # データ処理をシミュレート
    for i in range(1, 4):
        time.sleep(0.5)
        writer({"type": "売上レポート生成", "message": f"売上レポート生成進捗：{i *
                                                                             25}%"})
    writer({"type": "売上レポート生成", "message": "レポート生成完了"})
    return f"売上レポート：総収入150万元、前年同期比12%増"


@tool
def generate_inventory_report() -> str:
    """在庫レポートを生成する"""
    writer = get_stream_writer()
    writer("在庫分析を開始します...")
    time.sleep(0.5)
    writer("現在の在庫量を確認しています...")
    time.sleep(0.5)
    writer("在庫レポートを生成しています...")
    return "現在の在庫量は10000個で、在庫は十分であり異常はありません"


# レポート生成agentを作成
reporting_agent = create_agent(
    model=model,
    tools=[generate_sales_report, generate_inventory_report]
)
for chunk in reporting_agent.stream(
        {"messages": [{"role": "user", "content": "売上レポートと在庫レポートを生成してください"}]},
        stream_mode="custom"
):
    print(chunk)
    print("-" * 50)
```
