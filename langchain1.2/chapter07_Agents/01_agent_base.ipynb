## ツールをバインドして呼び出す方法
## カスタムツールをバインドする

```python
from langchain.chat_models import init_chat_model
import os
from rich import print
from langchain_core.tools import tool

OPENROUTER_API_KEY = os.getenv("OPENROUTER_API_KEY")
OPENROUTER_API_BASE = os.getenv("OPENROUTER_API_BASE")

model = init_chat_model(
    model="openai/gpt-4o-mini",
    model_provider="openai",
    api_key=OPENROUTER_API_KEY,
    base_url=OPENROUTER_API_BASE,
)


# ツールを定義
@tool
def getWeather(city: str):
    """
    天気情報を照会する
    
    Args:
        city:具体的な都市
    """
    return f"{city}は晴れ、気温30度です。"


agent = create_agent(
    model=model,
    tools=[getWeather, ]
)

response = agent.invoke({
    "messages": [
        {"role": "system",
         "content": "あなたは天気を調べるアシスタントです。ユーザーの質問に応じて天気を調べてください。天気と関係のない質問の場合は、その質問の答えは分からないと言ってください。"},
        {"role": "user", "content": "北京の天気はどうですか "},
    ]
})
print(response)
```

langchain内蔵のツールを呼び出す

```python
from langchain.chat_models import init_chat_model
import os
from rich import print
from langchain_core.tools import tool

OPENROUTER_API_KEY = os.getenv("OPENROUTER_API_KEY")
OPENROUTER_API_BASE = os.getenv("OPENROUTER_API_BASE")
TAVILY_API_KEY = os.getenv("TAVILY_API_KEY")

model = init_chat_model(
    model="openai/gpt-4o-mini",
    model_provider="openai",
    api_key=OPENROUTER_API_KEY,
    base_url=OPENROUTER_API_BASE,
)

# 内蔵ツール
web_search = TavilySearch(
    max_results=2,
    tavily_api_key=TAVILY_API_KEY
)

agent = create_agent(
    model=model,
    tools=[web_search, ]
)

response = agent.invoke({
    "messages": [{"role": "user", "content": "2024年のノーベル物理学賞受賞者が誰か調べてください。"},
                 ]
})
print(response["messages"][-1].content)
```

## 複数のツールをバインドして呼び出す

```python
from langchain.chat_models import init_chat_model
import os
from rich import print
from langchain_core.tools import tool

OPENROUTER_API_KEY = os.getenv("OPENROUTER_API_KEY")
OPENROUTER_API_BASE = os.getenv("OPENROUTER_API_BASE")
TAVILY_API_KEY = os.getenv("TAVILY_API_KEY")

model = init_chat_model(
    model="openai/gpt-4o-mini",
    model_provider="openai",
    api_key=OPENROUTER_API_KEY,
    base_url=OPENROUTER_API_BASE,
)

# 内蔵ツール
web_search = TavilySearch(
    max_results=2,
    tavily_api_key=TAVILY_API_KEY
)


# ツールを定義
@tool
def getWeather(city: str):
    """
    天気情報を照会する
    
    Args:
        city:具体的な都市
    """
    return f"{city}は晴れ、気温30度です。"


@tool(parse_docstring=True)
def get_news():
    """
    ニュース照会ツール
    """
    return "最近、世界的な半導体不足など複数の要因の影響を受け、各地の回収業者によると中古携帯電話の回収市場が“活況”を迎えており、回収価格は軒並み上昇し、中古携帯電話が“人気商品”となっている。"


agent = create_agent(
    model=model,
    tools=[getWeather, get_news, ]
)

response = agent.invoke({
    "messages": [{"role": "user", "content": "こんにちは、今日の南京の天気はどうですか？今日はどんなニュースがありますか？"},
                 ]
})
print(response)
```
