```python
from langchain.agents import create_agent
from langchain_core.messages import SystemMessage
from langchain_core.tools import tool
from rich import print


# ツール：2つの数の加算を実装
@tool
def add_numbers(a: int, b: int) -> str:
    """2つの数の和を計算して返す。"""
    return f"合計：{a + b}"


# カスタマーサポートアシスタントAgentを作成
agent = create_agent(
    model=model,
    tools=[add_numbers],  # ツールリスト
    # system_prompt="あなたは数学アシスタントです。日常の算数問題を解決します"
    system_prompt=SystemMessage(content="あなたは数学アシスタントです。日常の算数問題を解決します")
)
response = agent.invoke(
    {"messages": [
        {"role": "user", "content": "10に20を足してさらに30を足すといくつですか？"}
    ]},
)
print(response)
# print(response["messages"][-1].content)
```

```python
from langchain.agents import create_agent
from langchain.tools import tool
from langchain.messages import SystemMessage, HumanMessage

flag = 0


@tool
def get_weather(city: str):
    """
    天気照会ツール
    Args:
    city: 都市名
    """
    global flag
    flag += 1
    if flag < 3:
        # raise Exception("一時的にアクセスできません")
        return "TEMP_UNAVAILABLE: 天気サービスは一時的に利用できません。しばらくしてから再試行してください"
    return f"{city}は今日良い天気です"


messages = [
    HumanMessage("こんにちは、今日の杭州の天気はどうですか？")
]
agent = create_agent(
    model=model,
    tools=[get_weather],
    system_prompt=SystemMessage(
        "あなたは天気アシスタントです。"
        "ツールが 'TEMP_UNAVAILABLE:' で始まる結果を返した場合、"
        "それは一時的な故障を意味するので、すぐに諦めないでください。"
        "同じツールを再度呼び出し、最大3回まで再試行してください。"
        "3回試みても失敗する場合は、ユーザーにサービスが一時的に利用できないことを説明してください。"
    )
)
response = agent.invoke({"messages": messages})
# print(response)
for msg in response["messages"]:
    msg.pretty_print()
```
